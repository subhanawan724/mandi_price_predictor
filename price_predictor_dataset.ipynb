```python
import numpy as np
import pandas as pd


np.random.seed(42)
N_SAMPLES = 30000


mandis = [
    "Islamabad Sabzi Mandi",
    "Rawalpindi Raja Bazar",
    "Lahore Badami Bagh",
    "Faisalabad Grain Market",
    "Multan Goods Market",
    "Peshawar Central Mandi",
    "Gujranwala Fruit Market",
    "Quetta Satellite Town Mandi",
    "Hyderabad New Mandi",
    "Karachi Sabzi Mandi",
    "Sukkur Grain Market",
]


mandi_distance_ranges = {
    "Islamabad Sabzi Mandi": (150, 650),
    "Rawalpindi Raja Bazar": (150, 650),
    "Lahore Badami Bagh": (80, 500),
    "Faisalabad Grain Market": (50, 350),
    "Multan Goods Market": (30, 250),  
    "Peshawar Central Mandi": (180, 700),
    "Gujranwala Fruit Market": (60, 400),
    "Quetta Satellite Town Mandi": (200, 850),
    "Hyderabad New Mandi": (40, 300),
    "Karachi Sabzi Mandi": (150, 1100),  
    "Sukkur Grain Market": (40, 350),
}


catalog = {
    
    "Potato": ((30, 65), "Winter", "Low"),
    "Tomato": ((50, 130), "Winter", "High"),
    "Onion": ((55, 140), "Spring", "Low"),
    "Garlic (Local)": ((220, 380), "Winter", "Low"),
    "Garlic (China)": ((350, 550), "Winter", "Low"),
    "Ginger": ((380, 650), "Winter", "Low"),
    "Green Chilies": ((80, 220), "Summer", "High"),
    "Red Chili (Fresh)": ((120, 280), "Summer", "High"),
    "Cauliflower": ((40, 90), "Winter", "Medium"),
    "Cabbage": ((35, 80), "Winter", "Medium"),
    "Ladyfinger (Bhindi)": ((60, 150), "Summer", "High"),
    "Bitter Gourd (Karela)": ((70, 160), "Summer", "High"),
    "Bottle Gourd (Lauki)": ((40, 90), "Summer", "Medium"),
    "Ridge Gourd (Tori)": ((50, 110), "Summer", "High"),
    "Eggplant (Baingan)": ((40, 90), "Summer", "Medium"),
    "Spinach (Palak)": ((25, 60), "Winter", "High"),
    "Fenugreek (Methi)": ((30, 70), "Winter", "High"),
    "Mustard Greens (Sarson)": ((25, 55), "Winter", "High"),
    "Carrot": ((35, 80), "Winter", "Medium"),
    "Radish (Mooli)": ((25, 60), "Winter", "Medium"),
    "Turnip (Shalgam)": ((30, 70), "Winter", "Medium"),
    "Peas (Matar)": ((80, 220), "Winter", "High"),
    "Cucumber (Kheera)": ((40, 95), "Summer", "High"),
    "Zucchini": ((60, 130), "Summer", "Medium"),
    "Pumpkin (Halwa Kaddu)": ((30, 70), "Summer", "Low"),
    "Tinda": ((50, 120), "Summer", "High"),
    "Arbi (Taro Root)": ((70, 160), "Summer", "Medium"),
    "Capsicum (Shimla Mirch)": ((70, 180), "Winter", "High"),
    "Yellow Bell Pepper": ((180, 400), "Winter", "High"),
    "Red Bell Pepper": ((200, 420), "Winter", "High"),
    "Lemon (Desi)": ((100, 300), "Summer", "Medium"),
    "Lemon (China)": ((80, 200), "Summer", "Medium"),
    "Mint (Pudina)": ((20, 50), "Summer", "High"),
    "Coriander (Dhania)": ((30, 90), "Winter", "High"),
    "Spring Onion": ((40, 90), "Winter", "High"),
    "Beetroot (Chukandar)": ((50, 110), "Winter", "Medium"),
    "Sweet Potato (Shakarkandi)": ((40, 90), "Winter", "Low"),
    "Raw Banana (Kacha Kela)": ((40, 80), "Summer", "Medium"),
    "Drumstick (Sohanjna)": ((120, 280), "Spring", "Medium"),
    "Cluster Beans (Gwar)": ((60, 130), "Summer", "Medium"),
    "Raw Mango (Keri)": ((50, 110), "Summer", "Medium"),
    "Lotus Stem (Bheen)": ((80, 180), "Winter", "Low"),
    "Broccoli": ((120, 300), "Winter", "High"),
    "Lettuce": ((60, 150), "Winter", "High"),
    "Celery": ((100, 250), "Winter", "High"),
    "Zucchini (Yellow)": ((100, 220), "Summer", "Medium"),
    "Colocasia Leaves": ((40, 80), "Monsoon", "High"),
    "Mushroom (Button)": ((250, 500), "Winter", "High"),
    "Garlic Sprouts": ((100, 220), "Winter", "Medium"),
    "Green Chickpeas (Chholia)": ((100, 240), "Spring", "High"),
  
    "Apple (Kala Kulu)": ((180, 350), "Winter", "Low"),
    "Apple (Garcha)": ((130, 260), "Winter", "Low"),
    "Apple (Golden)": ((150, 290), "Winter", "Low"),
    "Banana (Dhaka)": ((100, 220), "Summer", "High"),
    "Banana (Sindhi)": ((70, 160), "Summer", "High"),
    "Mango (Chaunsa)": ((140, 300), "Summer", "High"),
    "Mango (Sindhri)": ((120, 260), "Summer", "High"),
    "Mango (Anwar Ratol)": ((160, 340), "Summer", "High"),
    "Mango (Dusehri)": ((110, 230), "Summer", "High"),
    "Orange (Kinnow)": ((80, 200), "Winter", "Medium"),
    "Orange (Fruiter)": ((100, 220), "Winter", "Medium"),
    "Mosambi (Sweet Lime)": ((90, 210), "Summer", "Medium"),
    "Grapefruit": ((70, 160), "Winter", "Medium"),
    "Guava (Amrood)": ((60, 140), "Winter", "High"),
    "Peach (Aroo)": ((130, 280), "Summer", "High"),
    "Plum (Aloo Bukhara)": ((150, 320), "Summer", "High"),
    "Apricot (Khubani)": ((160, 350), "Summer", "High"),
    "Grapes (Sundarkhani)": ((220, 450), "Summer", "High"),
    "Grapes (Gola)": ((150, 300), "Summer", "High"),
    "Grapes (Black)": ((200, 400), "Summer", "High"),
    "Pomegranate (Kandhari)": ((250, 550), "Winter", "Low"),
    "Pomegranate (Bedana)": ((300, 600), "Winter", "Low"),
    "Watermelon (Tarbooz)": ((25, 60), "Summer", "Medium"),
    "Melon (Kharbooza)": ((40, 95), "Summer", "Medium"),
    "Sweet Melon (Garma)": ((50, 110), "Summer", "Medium"),
    "Sarda": ((60, 130), "Summer", "Medium"),
    "Papaya": ((100, 220), "Summer", "High"),
    "Strawberry": ((180, 450), "Spring", "High"),
    "Falsa": ((150, 350), "Summer", "High"),
    "Jamun": ((140, 300), "Monsoon", "High"),
    "Lychee": ((250, 550), "Summer", "High"),
    "Persimmon (Amlok)": ((80, 180), "Winter", "Medium"),
    "Pear (Nashpati)": ((90, 200), "Summer", "Medium"),
    "Chiku (Sapodilla)": ((80, 170), "Summer", "High"),
    "Coconut (Fresh)": ((150, 300), "Summer", "Low"),
    "Dates (Aseel)": ((200, 450), "Summer", "Low"),
    "Dates (Rabai)": ((250, 500), "Summer", "Low"),
    "Fig (Anjeer Fresh)": ((200, 450), "Summer", "High"),
    "Mulberry (Shahtoot)": ((120, 280), "Spring", "High"),
    "Loquat": ((100, 220), "Spring", "High"),
    "Cherry": ((350, 800), "Summer", "High"),
    "Kiwi (Imported)": ((400, 850), "Winter", "Medium"),
    "Avocado": ((500, 1100), "Winter", "Medium"),
    "Dragon Fruit": ((600, 1300), "Summer", "Medium"),
    "Custard Apple (Sharifa)": ((180, 400), "Winter", "High"),
    "Wood Apple (Bael)": ((80, 180), "Summer", "Low"),
    "Sweet Lime (Malta)": ((100, 230), "Winter", "Medium"),
    "Green Apple": ((220, 420), "Winter", "Low"),
    "Pineapple": ((250, 550), "Summer", "Low"),
    "Wild Plum (Jaman)": ((120, 260), "Monsoon", "High"),
    "Rambutan (Imported)": ((700, 1500), "Summer", "High"),
    "Pistachio Fresh": ((600, 1200), "Summer", "Low"),
}

item_names = list(catalog.keys())
print(f"Total Unique Items Configured: {len(item_names)}")


seasons = ["Winter", "Summer", "Monsoon", "Spring"]

selected_items = np.random.choice(item_names, size=N_SAMPLES)
selected_mandis = np.random.choice(mandis, size=N_SAMPLES)
selected_seasons = np.random.choice(seasons, size=N_SAMPLES)


temp_array = []
rainfall_array = []

for season in selected_seasons:
    if season == "Winter":
        
        t = np.random.uniform(2.0, 20.0)
        r = np.random.uniform(0.0, 25.0)
    elif season == "Summer":
        
        t = np.random.uniform(30.0, 48.0)
        r = np.random.uniform(0.0, 40.0)
    elif season == "Monsoon":
        
        t = np.random.uniform(26.0, 38.0)
        r = np.random.uniform(40.0, 120.0)
    else:  
        
        t = np.random.uniform(18.0, 30.0)
        r = np.random.uniform(0.0, 30.0)

    temp_array.append(t)
    rainfall_array.append(r)


temp_array = np.array(temp_array)
rainfall_array = np.array(rainfall_array)

fuel_array = np.random.choice(
    [272.5, 278.0, 284.5, 292.0, 305.0], size=N_SAMPLES
)  
arrival_trucks = np.random.randint(4, 110, size=N_SAMPLES)
demand_index = np.random.choice([1, 2, 3, 4, 5], size=N_SAMPLES)  


distance_array = np.array([
    np.random.randint(
        mandi_distance_ranges[m][0], mandi_distance_ranges[m][1]
    )
    for m in selected_mandis
])


base_prices = []
perishability_list = []

for item, season in zip(selected_items, selected_seasons):
    (p_min, p_max), peak_season, perishability = catalog[item]
    perishability_list.append(perishability)

    price = np.random.randint(p_min, p_max)


    if season != peak_season:
        price = int(price * np.random.uniform(1.30, 1.50))

    base_prices.append(price)


df = pd.DataFrame({
    "Item": selected_items,
    "Perishability_Type": perishability_list,
    "Season": selected_seasons,
    "Mandi_Location": selected_mandis,
    "Wholesale_Base_Price": base_prices,
    "Temperature_C": np.round(temp_array, 1),
    "Rainfall_MM": np.round(rainfall_array, 1),
    "Fuel_Price_PKR": fuel_array,
    "Distance_KM": distance_array,
    "Arrival_Trucks": arrival_trucks,
    "Demand_Index": demand_index,
})



def calculate_spoilage_risk(row):
    score = 0

    
    if row["Perishability_Type"] == "High" and row["Temperature_C"] > 34.0:
        score += 2
    elif row["Temperature_C"] > 38.0:
        score += 1

    
    if row["Rainfall_MM"] > 50.0:
        score += 2
    elif row["Rainfall_MM"] > 20.0:
        score += 1

   
    if row["Distance_KM"] > 500:
        score += 1

    if score >= 3:
        return "High"
    elif score >= 1:
        return "Medium"
    return "Low"


df["Spoilage_Risk"] = df.apply(calculate_spoilage_risk, axis=1)


risk_multipliers = {"Low": 1.0, "Medium": 1.15, "High": 1.35}


def calculate_retail_price(row):
    base = row["Wholesale_Base_Price"]

    
    transport_cost = (row["Distance_KM"] * row["Fuel_Price_PKR"]) / 4800.0

    
    supply_ratio = row["Demand_Index"] / (row["Arrival_Trucks"] + 1)
    shortage_surge = supply_ratio * 4.5

    
    risk_mult = risk_multipliers[row["Spoilage_Risk"]]

    final_price = (base * risk_mult) + transport_cost + shortage_surge
    return round(final_price, 2)


df["Retail_Price_PKR"] = df.apply(calculate_retail_price, axis=1)


df.to_csv("pakistan_agri_102_items_30k.csv", index=False)
print("SUCCESS: Dataset generated with shape:", df.shape)
print(df[["Item", "Season", "Temperature_C", "Rainfall_MM", "Spoilage_Risk", "Retail_Price_PKR"]].head())
```

```python
import joblib
import pandas as pd
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.model_selection import train_test_split


df = pd.read_csv("pakistan_agri_102_items_30k.csv")

X1_raw = df[[
    "Perishability_Type",
    "Temperature_C",
    "Rainfall_MM",
    "Distance_KM",
]].copy()
y1 = df["Spoilage_Risk"]  


X1 = pd.get_dummies(X1_raw, columns=["Perishability_Type"], drop_first=False)


model1_features = list(X1.columns)


X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y1, test_size=0.2, random_state=42
)

model1 = RandomForestClassifier(n_estimators=100, random_state=42)
model1.fit(X1_train, y1_train)

acc1 = model1.score(X1_test, y1_test)
print(f"Model 1 Accuracy: {acc1 * 100:.2f}%")



print("\n--- Training Model 2 (Price Regressor) ---")


X2_raw = df[[
    "Item",
    "Mandi_Location",
    "Wholesale_Base_Price",
    "Fuel_Price_PKR",
    "Distance_KM",
    "Arrival_Trucks",
    "Demand_Index",
    "Spoilage_Risk",
]].copy()
y2 = df["Retail_Price_PKR"]  


X2 = pd.get_dummies(
    X2_raw,
    columns=["Item", "Mandi_Location", "Spoilage_Risk"],
    drop_first=False,
)


model2_features = list(X2.columns)


X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42
)

model2 = RandomForestRegressor(n_estimators=100, random_state=42)
model2.fit(X2_train, y2_train)

r2_score = model2.score(X2_test, y2_test)
print(f"Model 2 R² Score (Accuracy): {r2_score * 100:.2f}%")



joblib.dump(model1, "model1_spoilage_classifier.pkl")
joblib.dump(model2, "model2_price_regressor.pkl")


joblib.dump(model1_features, "model1_features.pkl")
joblib.dump(model2_features, "model2_features.pkl")

```
